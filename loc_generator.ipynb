```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('/content/project_Property2/model_trainer.csv')
```

```python
df['LOCALITY_WO_CITY'].unique()
```

```python
import requests, time, json

sectors = df['LOCALITY_WO_CITY'].unique()
headers = {"User-Agent": "GurgaonMapper/1.0"}
results = {}

for sector in sectors:
    r = requests.get(
        "https://nominatim.openstreetmap.org/search",
        params={"q": f"{sector}, Gurugram, Haryana, India",
                "format": "json", "limit": 1},
        headers=headers
    )
    data = r.json()
    if data:
        bb = data[0]["boundingbox"]  # [south, north, west, east]
        results[sector] = {
            "south": float(bb[0]), "north": float(bb[1]),
            "west":  float(bb[2]), "east":  float(bb[3]),
            "lat":   float(data[0]["lat"]),
            "lon":   float(data[0]["lon"]),
        }
        print(f"[OK] {sector}: SW({bb[0]},{bb[2]}) NE({bb[1]},{bb[3]})")
        results[sector] = f"SW({bb[0]},{bb[2]}) NE({bb[1]},{bb[3]})"
    time.sleep(1)

with open("sector_bbox.json", "w") as f:
    json.dump(results, f, indent=2)
```

```python
rows = []

for sector, value in results.items():
  row = {
        "Sector": sector,
        "cord": value,
    }
  rows.append(row)

df = pd.DataFrame(rows)
```

```python
df["SW"]=df['cord'].str.split(' ').str.get(0).str.replace('SW','').str.replace('(','').str.replace(')','')
df["NE"]=df['cord'].str.split(' ').str.get(1).str.replace('NE','').str.replace('(','').str.replace(')','')
```

```python
df.drop('cord',axis=1,inplace=True)
```
